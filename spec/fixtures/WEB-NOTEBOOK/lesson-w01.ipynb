## Digests

A hash maps any input to a fixed-length digest. For an $n$-bit output, a collision is expected after about

$$2^{n/2}$$

tries.

```python
import hashlib
hashlib.sha256(b"abc").hexdigest()
```

| n | tries |
|---|---|
| 128 | 2^64 |
| 256 | 2^128 |

````python
import hashlib
d = hashlib.sha256(b"abc").hexdigest()
assert len(d) == 64
print(d)
````
